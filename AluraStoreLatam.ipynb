# 2. Ventas por categoría

```python
category_counts = tienda.groupby('Categoría del Producto').size()
category_counts2 = tienda2.groupby('Categoría del Producto').size()
category_counts3 = tienda3.groupby('Categoría del Producto').size()
category_counts4 = tienda4.groupby('Categoría del Producto').size()
print(category_counts, category_counts2, category_counts3, category_counts4)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data preparation
categories = category_counts.index  # Product categories['Artículos para el hogar', 'Deportes y diversión', ..]
counts = [category_counts.values, category_counts2.values, category_counts3.values, category_counts4.values]
store_labels = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

x = np.arange(len(categories))  # Position of categories
width = 0.2  # Width of each bar

# Create grouped bars
fig, ax = plt.subplots(figsize=(12, 6))
for i in range(len(store_labels)):
    ax.bar(x + i * width, counts[i], width, label=store_labels[i])

# Formatting
ax.set_xlabel('Categoría del Producto')
ax.set_ylabel('Cantidad de Ventas')
ax.set_title('Ventas por Categoría en Diferentes Tiendas')
ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(categories, rotation=45)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

# 3. Calificación promedio de la tienda


```python
total_price_tienda = sum(tienda['Calificación'])
promedio_calificacion_tienda = total_price_tienda / len(tienda)
total_price_tienda2 = sum(tienda2['Calificación'])
promedio_calificacion_tienda2 = total_price_tienda2 / len(tienda2)
total_price_tienda3 = sum(tienda3['Calificación'])
promedio_calificacion_tienda3 = total_price_tienda3 / len(tienda3)
total_price_tienda4 = sum(tienda4['Calificación'])
promedio_calificacion_tienda4 = total_price_tienda4 / len(tienda4)

print(promedio_calificacion_tienda, promedio_calificacion_tienda2, promedio_calificacion_tienda3, promedio_calificacion_tienda4)
```

# 4. Productos más y menos vendidos

```python
product_counts_tienda = tienda.groupby('Producto').size().sort_values(ascending=False)
product_counts_tienda2 = tienda2.groupby('Producto').size().sort_values(ascending=False)
product_counts_tienda3 = tienda3.groupby('Producto').size().sort_values(ascending=False)
product_counts_tienda4 = tienda4.groupby('Producto').size().sort_values(ascending=False)

print(product_counts_tienda, product_counts_tienda2, product_counts_tienda3, product_counts_tienda4)
```
