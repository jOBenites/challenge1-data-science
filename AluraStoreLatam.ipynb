### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
#instalar biblioteca
!pip install matplotlib
```

```python
# Debes evaluar información como los ingresos, las categorías más vendidas, las reseñas de los clientes, los productos más vendidos y el envío promedio.
store_labels = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
```

#1. Análisis de facturación



```python
total_price_tienda = sum(tienda['Precio'])
total_price_tienda2 = sum(tienda2['Precio'])
total_price_tienda3 = sum(tienda3['Precio'])
total_price_tienda4 = sum(tienda4['Precio'])

print(total_price_tienda, total_price_tienda2, total_price_tienda3, total_price_tienda4)
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = [total_price_tienda, total_price_tienda2, total_price_tienda3, total_price_tienda4]

plt.figure() #Figura. Puede incluirse el tamaño con figsize
plt.bar(store_labels, data) #El gráfico
plt.title('Ventas por tienda') #El título
ax = plt.subplot() #Axis
ax.set_xticks(store_labels)  #Eje x
ax.set_xticklabels(store_labels) #Etiquetas del eje x
ax.set_xlabel('Tiendas')  #Nombre del eje x
ax.set_ylabel('Ventas')  #Nombre del eje y

plt.show()
```

# 2. Ventas por categoría

```python
category_counts = tienda.groupby('Categoría del Producto').size()
category_counts2 = tienda2.groupby('Categoría del Producto').size()
category_counts3 = tienda3.groupby('Categoría del Producto').size()
category_counts4 = tienda4.groupby('Categoría del Producto').size()
print(category_counts, category_counts2, category_counts3, category_counts4)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data preparation
categories = category_counts.index  # Product categories['Artículos para el hogar', 'Deportes y diversión', ..]
counts = [category_counts.values, category_counts2.values, category_counts3.values, category_counts4.values]

x = np.arange(len(categories))  # Position of categories
width = 0.2  # Width of each bar

# Create grouped bars
fig, ax = plt.subplots(figsize=(12, 6))
for i in range(len(store_labels)):
    ax.bar(x + i * width, counts[i], width, label=store_labels[i])

# Formatting
ax.set_xlabel('Categoría del Producto')
ax.set_ylabel('Cantidad de Ventas')
ax.set_title('Ventas por Categoría en Diferentes Tiendas')
ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(categories, rotation=45)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

# 3. Calificación promedio de la tienda


```python
total_price_tienda = sum(tienda['Calificación'])
promedio_calificacion_tienda = total_price_tienda / len(tienda)
total_price_tienda2 = sum(tienda2['Calificación'])
promedio_calificacion_tienda2 = total_price_tienda2 / len(tienda2)
total_price_tienda3 = sum(tienda3['Calificación'])
promedio_calificacion_tienda3 = total_price_tienda3 / len(tienda3)
total_price_tienda4 = sum(tienda4['Calificación'])
promedio_calificacion_tienda4 = total_price_tienda4 / len(tienda4)

print(promedio_calificacion_tienda, promedio_calificacion_tienda2, promedio_calificacion_tienda3, promedio_calificacion_tienda4)
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = [promedio_calificacion_tienda, promedio_calificacion_tienda2, promedio_calificacion_tienda3, promedio_calificacion_tienda4]

plt.figure() #Figura. Puede incluirse el tamaño con figsize
plt.plot(store_labels, data) #El gráfico
plt.title('Promedio de ventas por tienda') #El título
ax = plt.subplot() #Axis
ax.set_xticks(store_labels)  #Eje x
ax.set_xticklabels(store_labels) #Etiquetas del eje x
ax.set_xlabel('Tiendas')  #Nombre del eje x
ax.set_ylabel('Ventas')  #Nombre del eje y

plt.show()
```

# 4. Productos más y menos vendidos

```python
product_counts_tienda = tienda.groupby('Producto').size().sort_values(ascending=False)
product_counts_tienda2 = tienda2.groupby('Producto').size().sort_values(ascending=False)
product_counts_tienda3 = tienda3.groupby('Producto').size().sort_values(ascending=False)
product_counts_tienda4 = tienda4.groupby('Producto').size().sort_values(ascending=False)

print(product_counts_tienda, product_counts_tienda2, product_counts_tienda3, product_counts_tienda4)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Data preparation
categories = product_counts_tienda.index  # Product categories['Artículos para el hogar', 'Deportes y diversión', ..]
counts = [product_counts_tienda.values, product_counts_tienda2.values, product_counts_tienda3.values, product_counts_tienda4.values]
data = pd.DataFrame(counts, index=store_labels, columns=categories)

plt.figure(figsize=(12, 6))
sns.heatmap(data, annot=True, cmap="coolwarm", fmt="d")
plt.title('Distribución de Ventas por Categoría y Tienda')
plt.xlabel('Categoría del Producto')
plt.ylabel('Tiendas')


plt.show()
```

# 5. Envío promedio por tienda

```python
#OTRA FORMA DE SACAR EL PROMEDIO
average_shipping_tienda = tienda['Costo de envío'].mean()
average_shipping_tienda2 = tienda2['Costo de envío'].mean()
average_shipping_tienda3 = tienda3['Costo de envío'].mean()
average_shipping_tienda4 = tienda4['Costo de envío'].mean()

print(average_shipping_tienda, average_shipping_tienda2, average_shipping_tienda3, average_shipping_tienda4)
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = [average_shipping_tienda, average_shipping_tienda2, average_shipping_tienda3, average_shipping_tienda4]

plt.figure() #Figura. Puede incluirse el tamaño con figsize
plt.plot(store_labels, data) #El gráfico
plt.title('Envío promedio por tienda') #El título
ax = plt.subplot() #Axis
ax.set_xticks(store_labels)  #Eje x
ax.set_xticklabels(store_labels) #Etiquetas del eje x
ax.set_xlabel('Tiendas')  #Nombre del eje x
ax.set_ylabel('Envíos')  #Nombre del eje y

plt.show()
```
